```python
import os
import sys

# Select the backend before importing Keras. Restart the kernel after changing it.
if "keras" in sys.modules and sys.modules["keras"].backend.backend() != "torch":
    raise RuntimeError("Restart the kernel, then Run All to enable the PyTorch/MPS backend.")
os.environ["KERAS_BACKEND"] = "torch"

import numpy as np
import torch

device = "mps" if torch.backends.mps.is_available() else "cpu"
print(f"Keras backend: torch | Device: {device}")
if device == "cpu":
    print("MPS is unavailable in this Python process; using CPU.")
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
type(iris)
```

```python
print(iris.DESCR)
```

```python
# MPS requires float32 inputs instead of the Iris dataset's float64.
X = iris.data.astype(np.float32)
```

```python
X
```

```python
y = iris.target
```

```python
y
```

```python
# class 0 --> [1,0,0]
# class 1 --> [0,1,0]
# class 2 --> [0,0,1]
```

```python
# One-hot encode the Iris classes with NumPy.
# This avoids the Keras backend dependency for this preprocessing step.
```

```python
y = np.eye(len(iris.target_names), dtype=np.float32)[iris.target]
y.shape
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
# Keep the existing test set unchanged; reserve validation data from training only.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42, stratify=y_train.argmax(axis=1)
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
np.array([5,10,15,20])/20
```

```python
scaler_object = StandardScaler()
```

```python
scaler_object.fit(X_train)
```

```python
scaled_X_train = scaler_object.transform(X_train).astype(np.float32)
scaled_X_val = scaler_object.transform(X_val).astype(np.float32)
scaled_X_test = scaler_object.transform(X_test).astype(np.float32)
```

Restart the kernel and run all cells to select the PyTorch backend before Keras loads. On a Mac with available MPS support, Keras uses the Apple GPU. The first cell reports the selected device and falls back to CPU if MPS is unavailable.

```python
import keras
from keras import Input
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
from keras.callbacks import EarlyStopping
```

```python
# Reset the model and random seed so rerunning starts a reproducible training run.
keras.backend.clear_session()
keras.utils.set_random_seed(42)
model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(16, activation='relu'),
    Dense(16, activation='relu'),
    Dense(3, activation='softmax'),
])
model.to(device)
model.compile(loss='categorical_crossentropy', optimizer=Adam(learning_rate=0.01), metrics=['accuracy'])
```

```python
model.summary()
print("Model device:", next(model.parameters()).device)
```

The previous run reached 74% test accuracy and recognized only 2 of 15 versicolor examples. Standardize features using training data only, use a fixed initialization and a faster learning rate, and stop based on validation loss. The test set remains unchanged and is used only for the final evaluation.

```python
early_stopping = EarlyStopping(
    monitor='val_loss', patience=40, min_delta=0.0001, restore_best_weights=True
)
history = model.fit(
    scaled_X_train, y_train,
    validation_data=(scaled_X_val, y_val),
    epochs=400, batch_size=16, callbacks=[early_stopping], verbose=2
)
print(f"Training stopped after {len(history.history['loss'])} epochs.")
```

```python
model.predict(scaled_X_test)
```

```python
# Select the class with the highest predicted probability.
predicted_classes = model.predict(scaled_X_test).argmax(axis=1)
predicted_classes
```
