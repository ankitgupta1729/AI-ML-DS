```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(u'Tesla is looking at buying U.S. startup for $6 million')
```

```python
for token in doc:
    print(token.text)
```

```python
for token in doc:
    print(token.text,token.pos)
```

```python
for token in doc:
    print(token.text,token.pos_)
```

```python
for token in doc:
    print(token.text,token.pos_,token.dep_)
```

```python
nlp.pipe_names
```
