```python
@Language.component("set_custom_boundaries")
def set_custom_boundaries(doc):
    for token in doc[:-1]:
        if token.text == ";":
            doc[token.i + 1].is_sent_start = True
    return doc
```

```python
if "set_custom_boundaries" not in nlp.pipe_names:
    nlp.add_pipe("set_custom_boundaries", before="parser")
nlp.pipe_names
```

```python
# Register a custom component to split sentences on newlines.
from spacy.language import Language
```

```python
@Language.component("split_on_newlines")
def split_on_newlines(doc):
    start_sentence = True
    for token in doc:
        token.is_sent_start = start_sentence
        if "\n" in token.text:
            start_sentence = True
        elif not token.is_space:
            start_sentence = False
    return doc
```

```python
if "split_on_newlines" not in nlp.pipe_names:
    nlp.add_pipe("split_on_newlines", before="parser")
doc = nlp(mystring)
for sent in doc.sents:
    print(sent.text)
```
