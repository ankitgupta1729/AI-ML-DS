```python
text = "My telephone number are 777-555-1234 and 666-111-3456."

pattern = r'\d{3}-\d{3}-\d{4}'
all_phone_numbers = re.findall(pattern, text)
all_phone_numbers
```

```python
pattern = r'(\d{3})-(\d{3})-(\d{4})'
my_match = re.search(pattern, text)
my_match.group(0)
```

```python
my_match.group(1)
```

```python
my_match.group(2)
```

```python
my_match.group(3)
```
