```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
def show_ents(doc):
    if doc.ents:
        for ent in doc.ents:
            print(ent.text + " - " + ent.label_ + " - " + str(spacy.explain(ent.label_) ) )
    else:
        print("No entities found")
```

```python
doc = nlp(u"Hi, how are you")
```

```python
show_ents(doc)
```

```python
doc = nlp(u"May I go to Washington, D.C. next May to see the Washington Monument?")
```

```python
show_ents(doc)
```

```python
doc = nlp(u"Can I please have 500 dollars of Microsoft stock?")
```

```python
show_ents(doc)
```

```python
doc = nlp(u"Tesla to build a U.K. factory for $6 million.")
show_ents(doc)
```

```python
from spacy.tokens import Span
```

```python
ORG = doc.vocab.strings[u"ORG"]
ORG
```

```python
new_ent = Span(doc, 0, 1, label=ORG)
doc.ents = list(doc.ents) + [new_ent]
print([(ent.text, ent.label_) for ent in doc.ents])
```

```python
show_ents(doc)
```

```python
doc = nlp(u"Our company created a brand new vacuum cleaner. "
          u"This new vacumm-cleaner is the best in show.")
show_ents(doc)
```

```python
from spacy.matcher import PhraseMatcher
```

```python
matcher = PhraseMatcher(nlp.vocab)
phrase_list = [u"vacuum cleaner", u"vacumm-cleaner"]
phrase_patterns = [nlp(text) for text in phrase_list]
matcher.add(u"newproduct", None, *phrase_patterns)
```

```python
found_matches = matcher(doc)
print(found_matches)
```

```python
from spacy.tokens import Span
PROD = doc.vocab.strings[u"PRODUCT"]
new_ents = [Span(doc, match[1], match[2], label=PROD) for match in found_matches]
doc.ents = list(doc.ents) + new_ents

show_ents(doc)
```

```python
from spacy import displacy
```

```python
doc = nlp(u"Over the last quarter Apple sold nearly 20 thousand iPods for a profit of $6 million."
          u"By contrast, Sony only sold 8 thousand Walkman music players. ")
```

```python
displacy.render(doc, style="ent", jupyter=True)
```

```python
for sent in doc.sents:
    displacy.render(nlp(sent.text), style="ent", jupyter=True)
```

```python
options = {"ents": ['PRODUCT','ORG']}
displacy.render(doc, style="ent", jupyter=True,options=options)
```
