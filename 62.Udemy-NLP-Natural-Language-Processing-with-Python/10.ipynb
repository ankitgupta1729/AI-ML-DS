```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
from spacy.matcher import Matcher
```

```python
matcher = Matcher(nlp.vocab)
```

```python
#SolarPower
#Solar-Power
#Solar Power

pattern1 = [{'LOWER': 'solarpower'}]
pattern2 = [{'LOWER': 'solar'}, {'IS_PUNCT': True}, {'LOWER': 'power'}]
pattern3 = [{'LOWER': 'solar'}, {'LOWER': 'power'}]
```

```python
matcher.add("SolarPower", [pattern1, pattern2, pattern3])
```

```python
doc = nlp(u"The Solar Power industry continues to grow a solarpower increases. Solar-power is amazing.")
```

```python
found_matches = matcher(doc)
print(found_matches)
```

```python
for match_id, start, end in found_matches:
    string_id = nlp.vocab.strings[match_id]  # get string representation
    span = doc[start:end]                    # get the matched span
    print(match_id, string_id, start, end, span.text)
```

```python
matcher.remove("SolarPower")
```

```python
pattern1 = [{'LOWER': 'solarpower'}]
pattern2 = [{'LOWER': 'solar'}, {'IS_PUNCT': True, 'OP': '*'}, {'LOWER': 'power'}] 
```

```python
matcher.add("SolarPower", [pattern1, pattern2])
doc2 = nlp(u"Solar--power is solarpower yay!")
```

```python
found_matches = matcher(doc2)
print(found_matches)
```

```python
from spacy.matcher import PhraseMatcher
matcher = PhraseMatcher(nlp.vocab)
```

```python
with open('./Resources/UPDATED_NLP_COURSE/TextFiles/reaganomics.txt', encoding='cp1252') as f:
    doc3 = nlp(f.read())
```

```python
phrase_list = ['voodoo economics', 'supply-side economics', 'trickle-down economics', 'free-market economics']
phrase_patterns = [nlp(text) for text in phrase_list]
matcher.add("EconMatcher", phrase_patterns)
```

```python
found_matches = matcher(doc3)
print(found_matches)
```

```python
for match_id, start, end in found_matches:
    string_id = nlp.vocab.strings[match_id]  # get string representation
    span = doc3[start:end]                    # get the matched span
    print(match_id, string_id, start, end, span.text)
```
