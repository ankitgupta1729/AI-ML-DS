```python
doc = nlp("The quick brown fox jumped over the lazy dog's back.")
POS_counts = doc.count_by(spacy.attrs.POS)
print(POS_counts)
```

```python
doc.vocab[90].text
```

```python
for k,v in sorted(POS_counts.items()):
    print(f"{k:{5}}. {doc.vocab[k].text:{5}} {v}")
```

```python
TAG_counts = doc.count_by(spacy.attrs.TAG)
# print(TAG_counts)

for k,v in sorted(TAG_counts.items()):
    print(f"{k:{20}}. {doc.vocab[k].text:{5}} {v}")
```

```python
len(doc.vocab)
```

```python
DEP_counts = doc.count_by(spacy.attrs.DEP)
# print(DEP_counts)

for k,v in sorted(DEP_counts.items()):
    print(f"{k:{20}}. {doc.vocab[k].text:{5}} {v}")
```

```python
doc = nlp(u"The quick brown fox jumped over the lazy dog.")
print(doc.text)
```

```python
from spacy import displacy
displacy.render(doc, style="dep", jupyter=True)
```

```python
options = {"distance": 110, "compact": 'True', 'color': 'yellow', 'bg': '#09a3d5', 'font': 'Times'  }
displacy.render(doc, style="dep", jupyter=True, options=options)
```

```python
doc2 = nlp(u"This is a sentence. This is another sentence, possibly longer than the other.")
```

```python
spans = list(doc2.sents)
print(spans)
```

```python
displacy.serve(spans, style='dep', options={'distance': 110},auto_select_port=True)
```
