```python
import pandas as pd
# Load the data from a CSV file into a DataFrame
npr = pd.read_csv('./Resources/UPDATED_NLP_COURSE/05-Topic-Modeling/npr.csv')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm = tfidf.fit_transform(npr['Article'])
```

```python
dtm
```

```python
from sklearn.decomposition import NMF
# Create an instance of the NMF model with 7 topics
nmf_model = NMF(n_components=7, random_state=42)
```

```python
nmf_model.fit(dtm)
```

```python
tfidf.get_feature_names_out()[2300]
```

```python
for index,topic in enumerate(nmf_model.components_):
    print(f'The top 15 words for topic #{index}')
    print([tfidf.get_feature_names_out()[i] for i in topic.argsort()[-15:]])
    print('\n')
```

```python
topic_results = nmf_model.transform(dtm)
```

```python
topic_results[0]
```

```python
topic_results[0].argmax()
```

```python
topic_results.argmax(axis=1)
```

```python
npr['Topic'] = topic_results.argmax(axis=1)
npr.head()
```

```python
mytopic_dict = {0:'Health',1:'Election', 2: 'Legislation', 3: 'Politics', 4:'Election', 5:'Music', 6:'Music'}
npr['Topic Label'] = npr['Topic'].map(mytopic_dict)
npr.head()
```
