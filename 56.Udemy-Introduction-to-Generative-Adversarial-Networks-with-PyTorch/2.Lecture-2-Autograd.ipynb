```python
import torch
import numpy as np
import torch.nn.functional as F
```

`inputs` was a very simple, flexible example -- the four rows of a 2-bit
truth table -- and then we defined `outputs` as the ground truth, which is
`0, 1, 1, 0`. We create `weights` and set `.requires_grad` on that attribute
explicitly, or by passing it as a parameter to the constructor of a tensor --
now we use that second style for `bias`.

```python
inputs = torch.from_numpy(
            np.array(
                [
                    [0, 0],
                    [0, 1],
                    [1, 0],
                    [1, 1]
                ],
                dtype=np.float32
            )
        )

outputs = torch.from_numpy(
            np.array(
                [
                    0,
                    1,
                    1,
                    0
                ],
                dtype=np.float32
            )
        )

weights = torch.randn(1, 2)
weights.requires_grad = True

bias = torch.randn(1, requires_grad=True)
```

```python
F.linear(inputs, weights, bias)
```

```python
preds = F.linear(inputs, weights, bias)
loss = (outputs - preds).mean()
```

```python
preds.grad_fn
```

`grad_fn` is the function used to calculate the gradient for this tensor, so
we do the same but this time we save it against `loss`.

```python
loss.grad_fn
```

```python
weights.grad
```

We use `loss.backward()` -- this walks the autograd graph backward from
`loss` through `grad_fn`, accumulating gradients into every leaf tensor that
has `requires_grad=True` (here, `weights` and `bias`).

```python
loss.backward()
```

```python
weights.grad
```
