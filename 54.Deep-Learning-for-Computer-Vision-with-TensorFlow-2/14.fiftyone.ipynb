```python
import os
import sys
import logging

import tensorflow
import fiftyone as fo
import fiftyone.zoo as foz
from fiftyone import ViewField as F

# ---------------------------------------------------------------------
# Suppress the "show_confidence/show_index/show_label App config
# option has been removed..." warning flood.
#
# Root cause (confirmed against fiftyone/core/config.py): this WARNING
# fires every time an AppConfig is built from a dict containing those
# now-removed keys. Per the traceback's call stack
# (session/client.py:run_client -> subscribe -> Event.from_data ->
# StateDescription.from_dict -> AppConfig.from_dict), this is your
# already-open FiftyOne App browser tab repeatedly sending its config
# (cached in the browser from an older FiftyOne version) over its
# WebSocket connection -- once per state update, hence the flood. It
# is harmless (those options genuinely have no effect anymore either
# way), so this simply raises the log level to stop it cluttering
# output, rather than changing any actual behavior.
# ---------------------------------------------------------------------
logging.getLogger("fiftyone").setLevel(logging.ERROR)

print(tensorflow.__version__)
print(sys.version)

# ---------------------------------------------------------------------
# Local base directory for all downloads/exports, created in the
# current working directory (i.e. wherever this notebook lives).
#
# IMPORTANT FIX: earlier versions of this script passed `dataset_dir=`
# directly into foz.load_zoo_dataset(...), which is NOT a declared
# parameter of that function — it silently flows through **kwargs and,
# when combined with `classes=`/`label_types=` filtering, collides
# with a `dataset_dir` FiftyOne builds internally for the filtered
# download. That's exactly what caused:
#   TypeError: build_dataset_importer() got multiple values for
#   keyword argument 'dataset_dir'
#
# The correct, version-safe way to redirect where FiftyOne downloads
# zoo datasets is the global `fo.config.dataset_zoo_dir` setting below
# — set once, applies to every foz.load_zoo_dataset(...) call, no
# per-call `dataset_dir` kwarg needed (or wanted).
# ---------------------------------------------------------------------
BASE_DIR = os.path.abspath("./openimagesv6")
os.makedirs(BASE_DIR, exist_ok=True)
fo.config.dataset_zoo_dir = BASE_DIR

print(foz.list_zoo_datasets())

# ---------------------------------------------------------------------
# Data for classification
#
# drop_existing_dataset=True forces FiftyOne to rebuild the in-memory/
# on-disk *dataset object* fresh each time you re-run this cell,
# instead of reusing whatever was already loaded under this name (the
# "Loading existing dataset ... provide a custom dataset_name" message
# you saw). This does NOT re-download files from Google's servers —
# it just rebuilds FiftyOne's own dataset record from what's already
# on disk, so it's fast and safe to leave on.
# (Separately, `overwrite=True` would force re-downloading the actual
# image/label files themselves — not needed here since you already
# have them; only add that if you suspect the downloaded files are
# corrupted or incomplete.)
# ---------------------------------------------------------------------
if fo.dataset_exists("cls-open-images-v6"):
    fo.delete_dataset("cls-open-images-v6")

dataset_cls = foz.load_zoo_dataset(
    "open-images-v6",
    split="validation",
    label_types=["classifications"],
    classes=["Motorcycle", "Car"],
    max_samples=20,
    seed=2021,
    shuffle=True,
    dataset_name="cls-open-images-v6",
    drop_existing_dataset=True,
)

print(dataset_cls.get_field_schema())

# ---------------------------------------------------------------------
# Launch ONE App session and reuse it throughout the script by
# reassigning .dataset / .view, instead of calling fo.launch_app(...)
# repeatedly. Each fresh fo.launch_app(...) call spins up its own
# background WebSocket client thread; calling it 4 times in one script
# (as the original did) leaves old threads running after their cell's
# output stream has closed, which is what produced the second error
# you saw:
#   --- Logging error ---
#   ValueError: I/O operation on closed file
# Reusing a single `session` avoids that entirely.
# ---------------------------------------------------------------------
session = fo.launch_app(dataset_cls)

motor_cls = (
    dataset_cls.filter_labels("positive_labels",
        (F("label").is_in(["Motorcycle"]))))

car_cls = (
    dataset_cls.filter_labels("positive_labels",
        (F("label").is_in(["Car"]))))

session.view = motor_cls

# --- PLACEHOLDER: classification export ---
# motor_cls.export(
#     export_dir=os.path.join(BASE_DIR, "cls", "motor"),
#     dataset_type=fo.types.ImageClassificationDirectoryTree,
# )
# car_cls.export(
#     export_dir=os.path.join(BASE_DIR, "cls", "car"),
#     dataset_type=fo.types.ImageClassificationDirectoryTree,
# )

# ---------------------------------------------------------------------
# Data for detections
#
# IMPORTANT FIX: when label_types is a single type other than
# "classifications" (here, just ["detections"]), FiftyOne's Open
# Images importer yields one Detections object per sample rather than
# a dict of labels. Per FiftyOne's own documented behavior
# (Dataset.add_importer / import_samples), when the importer produces
# a single Label instance and no `label_field` is specified, it
# defaults to naming that field "ground_truth" — NOT "detections".
# That mismatch is exactly what caused:
#   ValueError: DatasetView has no field 'detections'
#
# Passing label_field="detections" explicitly below forces FiftyOne to
# store it under that name instead, matching the filter_labels(...)
# call further down.
#
# BELT-AND-SUSPENDERS: if you already ran the OLD (buggy) version of
# this cell once, FiftyOne may have a dataset named
# "det-open-images-v6" cached from that run, still using the old
# "ground_truth" field. drop_existing_dataset=True *should* rebuild it
# fresh either way, but the explicit fo.delete_dataset(...) call below
# removes any ambiguity — it's a no-op (silently does nothing) if no
# such dataset exists yet.
# ---------------------------------------------------------------------
if fo.dataset_exists("det-open-images-v6"):
    fo.delete_dataset("det-open-images-v6")

dataset_det = foz.load_zoo_dataset(
    "open-images-v6",
    split="validation",
    label_types=["detections"],
    label_field="detections",
    classes=["Motorcycle", "Car"],
    max_samples=20,
    dataset_name="det-open-images-v6",
    drop_existing_dataset=True,
)

# Sanity check: print the actual field schema so you can SEE the field
# name directly, rather than assuming — look for "detections" in this
# output before running filter_labels() below. If you ever hit a
# similar "DatasetView has no field '...'" error again for a different
# dataset, run `print(dataset.get_field_schema())` first to find the
# real field name rather than guessing.
print(dataset_det.get_field_schema())

session.dataset = dataset_det

motor_car_det = (
    dataset_det
    .filter_labels(
        "detections",
        (F("label").is_in(["Motorcycle", "Car"]))))

session.view = motor_car_det

# ---------------------------------------------------------------------
# Export detections — YOLOv4 format
# ---------------------------------------------------------------------
motor_car_det.export(
    export_dir=os.path.join(BASE_DIR, "det", "Yolov4"),
    dataset_type=fo.types.YOLOv4Dataset,
)

# ---------------------------------------------------------------------
# Export detections — TFRecords format (TF Object Detection API)
# ---------------------------------------------------------------------
motor_car_det.export(
    export_dir=os.path.join(BASE_DIR, "det", "TFObjectDection"),
    dataset_type=fo.types.TFObjectDetectionDataset,
)

# References:
# https://storage.googleapis.com/openimages/web/index.html
# https://voxel51.com/docs/fiftyone/tutorials/open_images.html
# https://voxel51.com/docs/fiftyone/user_guide/export_datasets.html
```
