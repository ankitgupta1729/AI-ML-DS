**addition**

```python
t1 = torch.rand(2,3)
t2 = torch.rand(2,3)

print(t1)
print(t2)

print("Addition: \n",t1+t2)
print("Subtraction: \n",t1-t2)
print("Multiplication: \n",t1*t2)
print("Division: \n",t1/t2)
```

```python
print(t1.add(t2))
print(torch.add(t1, t2))

print(t1.sub(t2))
print(torch.sub(t1, t2))

print(t1.mul(t2))
print(torch.mul(t1, t2))

print(t1.div(t2))
print(torch.div(t1, t2))
```

**average**

```python
print(torch.mean(t1))
print(t1.mean())
```
