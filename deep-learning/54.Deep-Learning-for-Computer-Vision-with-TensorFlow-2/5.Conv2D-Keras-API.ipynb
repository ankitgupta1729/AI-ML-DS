```python
import tensorflow as tf
from tensorflow.keras.models import Model
```

#### Convolution filter

```python
input_shape = (1, 28, 28, 3)
x = tf.random.normal(input_shape)
y = tf.keras.layers.Conv2D(filters=2, kernel_size= 5, strides=(1,1), activation='relu', input_shape=input_shape[1:])(x)
print(y.shape)
```

#### Convolution with a filter size as the size of the image

```python
input_shape = (1, 28, 28, 3)
x = tf.random.normal(input_shape)
y = tf.keras.layers.Conv2D(filters=2, kernel_size= 28, strides=(1,1), activation='relu', input_shape=input_shape[1:])(x)
print(y.shape)
```

#### Padding "valid"

```python
input_shape = (1, 28, 28, 3)
x = tf.random.normal(input_shape)
y = tf.keras.layers.Conv2D(filters=2,kernel_size= 5,strides=(1,1), activation='relu', padding="valid", input_shape=input_shape[1:])(x)
print(y.shape)
```

```python
input_shape = (1, 28, 28, 3)
x = tf.random.normal(input_shape)
y = tf.keras.layers.Conv2D(filters=2,kernel_size= 5, strides=(1,1), activation='relu', padding="same", input_shape=input_shape[1:])(x)
print(y.shape)
```

```python
input_shape = (1, 28, 28, 3)
x = tf.random.normal(input_shape)
y = tf.keras.layers.Conv2D(filters=2,kernel_size= 5, strides=(1,1),dilation_rate=2, 
                           activation='relu', input_shape=input_shape[1:])(x)
print(y.shape)
```

```python
input_shape = (1, 28, 28, 3)
x = tf.random.normal(input_shape)
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(1,3,activation='relu',name='conv2d_1',input_shape=input_shape[1:]),
    tf.keras.layers.Conv2D(2,5,activation='relu',name='conv2d_2')
])
output = model(x)
print(output.shape)
```

```python
for layer in model.layers:
    print(layer.name,layer)
```
