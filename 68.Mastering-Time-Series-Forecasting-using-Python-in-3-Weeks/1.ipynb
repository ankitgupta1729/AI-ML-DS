## Control Flow Statements
```
if <condition>:
    <statements>
elif <condition>:
    <statements>

...


else:
    <statements>
```

````python
a=10
b=20
if a>b:
    print('{} is greater than {}'.format(a,b))
elif a<b:
    print('{} is less than {}'.format(a,b))
else:
    print('{} is equal to {}'.format(a,b))
````

````python
a = float(input('Enter the value of a='))
b = float(input('Enter the value of b='))
if a>b:
    print('{} is greater than {}'.format(a,b))
elif a<b:
    print('{} is less than {}'.format(a,b))
else:
    print('{} is equal to {}'.format(a,b))
````
