Let's start with defining a list, one of the common objects where one can iterate on:

```python
list_example = [10,20,30,40]
```

```python
for element in list_example:
    print(element*2)
```

```python
for element in list_example:
    print(element*4)
```

```python
for index, element in enumerate(list_example):
    print(index, element)
```

```python
example_dict = {'January':1,'February':2,'March':3}

for month, item in example_dict.items():
    print('{} is month number {}'.format(month, item))
```

### If Statements

Other than loops, we can also use conditionals to do some cool stuff in the language:

```python
for element in [True, True, False]:
    if element == False:
        print('Got a False!')
```

```python
for element in [True, True, False]:
    if not(element):
        print('Got a False!')
```

```python
if 'a' in 'bc':
    print('found and a!')
else:
    print('did not found an a!')
```

### While Statements

`While` statements are another interesting statement we can use in Python. Particularly, these `while` statements are very prone to infinite loops, if we don't update the variable we are testing on the condition:

```python
takeoff = 10

i = 0
while i < takeoff:
    print('Countdown: {}'.format(i))
```

While does not increment automatically like **for**!

```python
takeoff = 10

i = 0
while i < takeoff:
    print('Countdown: {}'.format(i))
    i += 1
```
